# STEP 4: LOAD AND CLEAN DATA


```python

def load_and_clean_data(file_name, uploaded):
    """Load Excel data and perform comprehensive cleaning"""

    print("\n Loading and cleaning data...")

    # Read the uploaded file directly
    df = pd.read_excel(file_name, sheet_name='Sheet1')

    print(f"✓ Loaded {len(df)} raw records")

    # Standardize column names
    df.columns = df.columns.str.lower().str.strip()

    # Remove empty rows
    df = df.dropna(subset=['rn'], how='all')
    df = df[df['year'].notna()]

    # Convert year to integer
    df['year'] = df['year'].astype(int)

    # Clean microorganism names
    df['microorganisms'] = df['microorganisms'].str.lower().str.strip()
    microbe_mapping = {
        'staphylococcus aureus': 'Staphylococcus aureus',
        'pseudomonas aeruginosa': 'Pseudomonas aeruginosa',
        'klebsiella pneumonia': 'Klebsiella pneumoniae',
        'enterococcus': 'Enterococcus spp.',
        'escherichia coli': 'Escherichia coli',
        'proteus mirabilis': 'Proteus mirabilis',
        'serratia marcescens': 'Serratia marcescens',
        'bacillus': 'Bacillus spp.',
        'enterobacter': 'Enterobacter spp.',
        'acinobacter': 'Acinetobacter spp.',
        'streptococcus beta-hemolytic group a': 'Streptococcus pyogenes',
        'bacteroides fragilis': 'Bacteroides fragilis',
        'corynebacterium': 'Corynebacterium spp.',
        'enterobacter (carbapenem resistant)': 'CRE Enterobacter',
        'entrobacter': 'Enterobacter spp.'
    }
    df['microorganisms_clean'] = df['microorganisms'].replace(microbe_mapping)

    # Create binary variables for comorbidities
    comorbidity_cols = ['dm', 'hpt', 'ckd', 'cardiac disease']
    for col in comorbidity_cols:
        if col in df.columns:
            df[f'{col}_binary'] = df[col].map({'yes': 1, 'no': 0, '': 0, np.nan: 0})
        else:
            df[f'{col}_binary'] = 0

    # Create binary for smoking
    df['smoking_binary'] = df['smoking'].map({'yes': 1, 'no': 0, '': 0, np.nan: 0})

    # Convert surgery duration to numeric
    df['surgery_duration'] = pd.to_numeric(df['surgery duration (hours)'], errors='coerce')
    df['surgery_duration'] = df['surgery_duration'].fillna(df['surgery_duration'].median())

    # Convert preoperative duration to numeric
    df['preop_duration'] = pd.to_numeric(df['preoperative duration (days)'], errors='coerce')
    df['preop_duration'] = df['preop_duration'].fillna(0)

    # Create age groups
    df['age'] = pd.to_numeric(df['age'], errors='coerce')
    df['age_group'] = pd.cut(df['age'], bins=[0, 18, 40, 60, 100],
                              labels=['<18 years', '18-40 years', '41-60 years', '>60 years'])

    # Create surgery duration groups
    df['surgery_group'] = pd.cut(df['surgery_duration'],
                                  bins=[0, 2, 3, 10],
                                  labels=['<2 hours', '2-3 hours', '>3 hours'])

    # Create year groups for trend analysis
    df['year_group'] = pd.cut(df['year'], bins=[2013, 2016, 2019, 2023],
                               labels=['2014-2016', '2017-2019', '2020-2023'])

    # Calculate resistance count
    antibiotic_cols = ['ampicillin', 'gentamicin', 'penicillin g', 'cloxacillin',
                      'erythromycin', 'trimethoprim sulfamethoxazole',
                      'amoxycillin-clavulanate', 'cefuroxime',
                      'piperacillin-tazobactam', 'ciprofloxacin', 'meropenem']

    resistance_cols_list = []
    for col in antibiotic_cols:
        if col in df.columns:
            df[f'{col}_resistant'] = (df[col] == 'no').astype(int)
            resistance_cols_list.append(f'{col}_resistant')

    if resistance_cols_list:
        df['resistance_count'] = df[resistance_cols_list].sum(axis=1)
        df['mdro'] = (df['resistance_count'] >= 3).astype(int)
    else:
        df['resistance_count'] = 0
        df['mdro'] = 0

    print(f"✓ Data cleaning completed")
    print(f"✓ Final dataset: {len(df)} patients")
    print(f"✓ Years covered: {df['year'].min()} - {df['year'].max()}")

    return df

# Load and clean data
df = load_and_clean_data(file_name, uploaded)

# Display first few rows
print("\n First 5 rows of cleaned data:")
display(df.head())

```

# SECTION 6: TEMPORAL TREND ANALYSIS

```python
# ============================================
# SECTION 6: TEMPORAL TREND ANALYSIS
# ============================================

print("\n" + "="*80)
print("SECTION 6: TEMPORAL TREND ANALYSIS")
print("="*80)
print("🎯 SPECIFIC OBJECTIVE 1: To describe the pattern of reported cases")
print("📋 RESEARCH QUESTION 1: How has the pattern changed over years?")
print("="*80)

def temporal_trend_analysis(df):
    """Analyze temporal patterns and trends over 10 years (descriptive only)"""

    # Yearly case counts
    yearly_counts = df.groupby('year').size().reset_index(name='count')
    yearly_counts['cumulative'] = yearly_counts['count'].cumsum()
    yearly_counts['percentage'] = (yearly_counts['count'] / yearly_counts['count'].sum()) * 100
    yearly_counts['yoy_change'] = yearly_counts['count'].pct_change() * 100

    # Descriptive statistics only (no linear regression)
    total_cases = yearly_counts['count'].sum()
    mean_cases = yearly_counts['count'].mean()
    median_cases = yearly_counts['count'].median()
    min_cases = yearly_counts['count'].min()
    max_cases = yearly_counts['count'].max()
    peak_year = yearly_counts.loc[yearly_counts['count'].idxmax(), 'year']

    # Simple trend direction (increase or decrease from first to last year)
    first_year_cases = yearly_counts.iloc[0]['count']
    last_year_cases = yearly_counts.iloc[-1]['count']

    if last_year_cases > first_year_cases:
        trend_direction = "Increasing"
    elif last_year_cases < first_year_cases:
        trend_direction = "Decreasing"
    else:
        trend_direction = "Stable"

    print(f"\n📈 Temporal Trend Results (Descriptive):")
    print(f"  Total cases (2014-2023): {total_cases}")
    print(f"  Mean annual cases: {mean_cases:.1f} (±{yearly_counts['count'].std():.1f})")
    print(f"  Median annual cases: {median_cases:.0f}")
    print(f"  Range: {min_cases} - {max_cases} cases/year")
    print(f"  Peak year: {peak_year} ({max_cases} cases)")
    print(f"  First year (2014): {first_year_cases} cases")
    print(f"  Last year (2023): {last_year_cases} cases")
    print(f"  Overall trend direction: {trend_direction}")

    print("\n📅 Yearly Case Distribution:")
    print(yearly_counts[['year', 'count', 'cumulative', 'percentage']].to_string(index=False))

    # Create figures (WITHOUT trend line)
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    # 1. Annual cases (NO trend line)
    ax1 = axes[0, 0]
    ax1.bar(yearly_counts['year'], yearly_counts['count'], alpha=0.7, color='steelblue',
            label='Annual cases', edgecolor='black')
    ax1.plot(yearly_counts['year'], yearly_counts['count'], 'o-', color='darkred',
             linewidth=2, markersize=8, label='Observed')
    ax1.set_xlabel('Year', fontsize=11)
    ax1.set_ylabel('Number of FRI Cases', fontsize=11)
    ax1.set_title('Annual FRI Cases (2014-2023)', fontsize=12, fontweight='bold')
    ax1.legend(loc='upper left')
    ax1.grid(True, alpha=0.3)
    ax1.set_xticks(range(2014, 2024))

    # 2. Cumulative cases
    ax2 = axes[0, 1]
    ax2.fill_between(yearly_counts['year'], yearly_counts['cumulative'], alpha=0.3, color='steelblue')
    ax2.plot(yearly_counts['year'], yearly_counts['cumulative'], 'o-', color='darkblue',
             linewidth=2, markersize=8)
    ax2.set_xlabel('Year', fontsize=11)
    ax2.set_ylabel('Cumulative Cases', fontsize=11)
    ax2.set_title('Cumulative FRI Cases Over Time', fontsize=12, fontweight='bold')
    ax2.grid(True, alpha=0.3)
    ax2.set_xticks(range(2014, 2024))

    # 3. Year-over-year percentage change
    ax3 = axes[1, 0]
    colors_yoy = ['green' if x >= 0 else 'red' for x in yearly_counts['yoy_change'][1:]]
    ax3.bar(yearly_counts['year'][1:], yearly_counts['yoy_change'][1:],
            color=colors_yoy, alpha=0.7, edgecolor='black')
    ax3.axhline(y=0, color='black', linestyle='-', linewidth=1)
    ax3.set_xlabel('Year', fontsize=11)
    ax3.set_ylabel('Year-over-Year Change (%)', fontsize=11)
    ax3.set_title('Annual Percentage Change in FRI Cases', fontsize=12, fontweight='bold')
    ax3.grid(True, alpha=0.3, axis='y')
    ax3.set_xticks(range(2015, 2024))

    # 4. Percentage distribution by year
    ax4 = axes[1, 1]
    ax4.plot(yearly_counts['year'], yearly_counts['percentage'], 'o-', color='purple',
             linewidth=2, markersize=8)
    ax4.fill_between(yearly_counts['year'], yearly_counts['percentage'], alpha=0.3, color='purple')
    ax4.set_xlabel('Year', fontsize=11)
    ax4.set_ylabel('Percentage of Total Cases (%)', fontsize=11)
    ax4.set_title('Annual Percentage Contribution', fontsize=12, fontweight='bold')
    ax4.grid(True, alpha=0.3)
    ax4.set_xticks(range(2014, 2024))

    plt.tight_layout()
    plt.savefig('Figure_2_Temporal_Trends.png', dpi=300, bbox_inches='tight')
    plt.show()

    print("\n✅ SPECIFIC OBJECTIVE 1 COMPLETED: Pattern of reported cases described")
    print(f"✅ RESEARCH QUESTION 1 ANSWERED: {trend_direction} trend observed from 2014 to 2023")

    return yearly_counts

trend_results = temporal_trend_analysis(df)
```

```python
# ============================================
# SECTION 7: RISK FACTOR ANALYSIS
# ============================================

print("\n" + "="*80)
print("SECTION 7: RISK FACTOR ANALYSIS")
print("="*80)
print("🎯 SPECIFIC OBJECTIVE 2: To determine the predisposing factors")
print("📋 RESEARCH QUESTION 2: What risk factors/comorbid conditions contribute to FRI?")
print("="*80)

def risk_factor_analysis(df):
    """Comprehensive risk factor analysis"""

    risk_factors = {
        'Age >60 years': (df['age'] > 60).astype(int),
        'Male gender': (df['gender'] == 'male').astype(int),
        'Smoking': df['smoking_binary'],
        'Diabetes mellitus (DM)': df['dm_binary'] if 'dm_binary' in df.columns else pd.Series([0]*len(df)),
        'Hypertension (HPT)': df['hpt_binary'] if 'hpt_binary' in df.columns else pd.Series([0]*len(df)),
        'Chronic kidney disease (CKD)': df['ckd_binary'] if 'ckd_binary' in df.columns else pd.Series([0]*len(df)),
        'Cardiac disease': df['cardiac disease_binary'] if 'cardiac disease_binary' in df.columns else pd.Series([0]*len(df)),
        'Surgery >3 hours': (df['surgery_duration'] > 3).astype(int),
        'Preoperative stay >7 days': (df['preop_duration'] > 7).astype(int),
        'Intramedullary nails': (df['intramedullary nails'] == 'yes').astype(int),
        'Plating': (df['plating'] == 'yes').astype(int)
    }

    risk_results = []
    for factor_name, factor_values in risk_factors.items():
        n_total = len(factor_values.dropna())
        n_positive = factor_values.sum()
        percentage = (n_positive / n_total) * 100 if n_total > 0 else 0
        se = np.sqrt(percentage * (100 - percentage) / n_total) if n_total > 0 else 0
        ci_lower = percentage - 1.96 * se
        ci_upper = percentage + 1.96 * se

        risk_results.append({
            'Risk Factor': factor_name,
            'N with risk': n_positive,
            'Total N': n_total,
            'Prevalence (%)': f"{percentage:.1f}",
            '95% CI': f"({ci_lower:.1f}-{ci_upper:.1f})"
        })

    risk_df = pd.DataFrame(risk_results)
    print("\n📊 Prevalence of Risk Factors:")
    print(risk_df.to_string(index=False))

    # Identify top 3 risk factors
    risk_df['Prevalence_num'] = risk_df['Prevalence (%)'].str.replace('%', '').astype(float)
    top_risks = risk_df.nlargest(3, 'Prevalence_num')[['Risk Factor', 'Prevalence (%)']]

    print("\n🔍 TOP 3 PREDISPOSING FACTORS:")
    for idx, row in top_risks.iterrows():
        print(f"  {idx+1}. {row['Risk Factor']}: {row['Prevalence (%)']}")

    # Create bar chart
    fig, ax = plt.subplots(figsize=(12, 8))
    plot_data = risk_df.copy()
    plot_data['Prevalence'] = plot_data['Prevalence (%)'].str.replace('%', '').astype(float)
    plot_data = plot_data.sort_values('Prevalence', ascending=True)

    bars = ax.barh(plot_data['Risk Factor'], plot_data['Prevalence'],
                   color='coral', alpha=0.7, edgecolor='black')
    ax.set_xlabel('Prevalence (%)', fontsize=11)
    ax.set_title('Prevalence of Risk Factors in FRI Patients', fontsize=12, fontweight='bold')
    ax.grid(True, alpha=0.3, axis='x')

    for i, (bar, val) in enumerate(zip(bars, plot_data['Prevalence'])):
        ax.text(val + 1, bar.get_y() + bar.get_height()/2, f'{val:.1f}%',
                va='center', fontsize=9, fontweight='bold')

    plt.tight_layout()
    plt.savefig('Figure_3_Risk_Factors.png', dpi=300, bbox_inches='tight')
    plt.show()

    print("\n✅ SPECIFIC OBJECTIVE 2 COMPLETED: Predisposing factors identified")
    print("✅ RESEARCH QUESTION 2 ANSWERED: Top risk factors listed above")

    return risk_df, top_risks

risk_results, top_risks = risk_factor_analysis(df)

```

```python
# ============================================
# SECTION 8: MICROBIOLOGICAL ANALYSIS
# ============================================

print("\n" + "="*80)
print("SECTION 8: MICROBIOLOGICAL ANALYSIS")
print("="*80)
print("🎯 SPECIFIC OBJECTIVE 3: To describe type of organisms")
print("📋 RESEARCH QUESTION 3: What are the most common microorganisms?")
print("="*80)

def microbiological_analysis(df):
    """Comprehensive microbiological analysis"""

    microbe_df = df[df['microorganisms_clean'].notna() & (df['microorganisms_clean'] != '')]

    if len(microbe_df) == 0:
        print("No microorganism data available")
        return None

    microbe_counts = microbe_df['microorganisms_clean'].value_counts()
    microbe_percentages = (microbe_counts / len(microbe_df)) * 100

    microbe_summary = pd.DataFrame({
        'Microorganism': microbe_counts.index,
        'N': microbe_counts.values,
        'Percentage (%)': microbe_percentages.values
    })

    print("\n🦠 Microbial Distribution:")
    print(microbe_summary.to_string(index=False))

    print("\n🔝 TOP 5 MOST COMMON MICROORGANISMS:")
    for idx, row in microbe_summary.head(5).iterrows():
        print(f"  {idx+1}. {row['Microorganism']}: {row['N']} ({row['Percentage (%)']:.1f}%)")

    # Create figures
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    # 1. Pie chart
    ax1 = axes[0, 0]
    colors = sns.color_palette('husl', n_colors=min(len(microbe_summary), 10))
    ax1.pie(microbe_summary['N'].head(8), labels=microbe_summary['Microorganism'].head(8),
            autopct='%1.1f%%', colors=colors, startangle=90)
    ax1.set_title('Distribution of Isolated Pathogens', fontsize=12, fontweight='bold')

    # 2. Bar chart
    ax2 = axes[0, 1]
    top_microbes = microbe_summary.head(8)
    bars = ax2.barh(range(len(top_microbes)), top_microbes['N'],
                    color=colors[:len(top_microbes)], edgecolor='black')
    ax2.set_yticks(range(len(top_microbes)))
    ax2.set_yticklabels(top_microbes['Microorganism'])
    ax2.set_xlabel('Number of Isolates', fontsize=11)
    ax2.set_title('Top 8 Microorganisms Causing FRI', fontsize=12, fontweight='bold')
    ax2.grid(True, alpha=0.3, axis='x')

    for i, (bar, val) in enumerate(zip(bars, top_microbes['N'])):
        ax2.text(val + 0.5, bar.get_y() + bar.get_height()/2, f'n={val}',
                va='center', fontsize=9, fontweight='bold')

    # 3. Trend of top pathogens
    ax3 = axes[1, 0]
    top_3_pathogens = microbe_summary.head(3)['Microorganism'].tolist()
    for pathogen in top_3_pathogens:
        pathogen_data = df[df['microorganisms_clean'] == pathogen]
        yearly_pathogen = pathogen_data.groupby('year').size()
        years_range = range(2014, 2024)
        counts = [yearly_pathogen.get(year, 0) for year in years_range]
        ax3.plot(years_range, counts, 'o-', linewidth=2, markersize=6, label=pathogen)
    ax3.set_xlabel('Year', fontsize=11)
    ax3.set_ylabel('Number of Isolates', fontsize=11)
    ax3.set_title('Trend of Common Pathogens (2014-2023)', fontsize=12, fontweight='bold')
    ax3.legend(loc='best')
    ax3.grid(True, alpha=0.3)
    ax3.set_xticks(range(2014, 2024))

    # 4. Heatmap
    ax4 = axes[1, 1]
    pathogen_year_matrix = pd.crosstab(df['microorganisms_clean'], df['year'])
    pathogen_year_matrix = pathogen_year_matrix.loc[pathogen_year_matrix.sum(axis=1).sort_values(ascending=False).index[:10]]
    im = ax4.imshow(pathogen_year_matrix.values, cmap='YlOrRd', aspect='auto', interpolation='nearest')
    ax4.set_xticks(range(len(pathogen_year_matrix.columns)))
    ax4.set_xticklabels(pathogen_year_matrix.columns, rotation=45)
    ax4.set_yticks(range(len(pathogen_year_matrix.index)))
    ax4.set_yticklabels(pathogen_year_matrix.index, fontsize=8)
    ax4.set_xlabel('Year', fontsize=11)
    ax4.set_ylabel('Pathogen', fontsize=11)
    ax4.set_title('Pathogen Distribution by Year', fontsize=12, fontweight='bold')
    plt.colorbar(im, ax=ax4, label='Number of Isolates')

    plt.tight_layout()
    plt.savefig('Figure_4_Microbial_Distribution.png', dpi=300, bbox_inches='tight')
    plt.show()

    print("\n✅ SPECIFIC OBJECTIVE 3 (Part 1) COMPLETED: Types of organisms described")
    print("✅ RESEARCH QUESTION 3 ANSWERED: Most common microorganisms identified")

    return microbe_summary

microbe_results = microbiological_analysis(df)

```

```python
# ============================================
# SECTION 9: ANTIBIOTIC TREATMENT PATTERN ANALYSIS (CORRECTED)
# ============================================

print("\n" + "="*80)
print("SECTION 9: ANTIBIOTIC TREATMENT PATTERN ANALYSIS")
print("="*80)
print("🎯 SPECIFIC OBJECTIVE 3: To describe treatment given")
print("📋 RESEARCH QUESTION 4: What are the antibiotic treatment patterns?")
print("="*80)

def antibiotic_treatment_analysis(df):
    """Analyze antibiotic treatment patterns (NOT susceptibility)"""

    # Define antibiotics of interest
    antibiotics = {
        'ampicillin': 'Ampicillin',
        'gentamicin': 'Gentamicin',
        'penicillin g': 'Penicillin G',
        'cloxacillin': 'Cloxacillin',
        'erythromycin': 'Erythromycin',
        'cefuroxime': 'Cefuroxime',
        'ciprofloxacin': 'Ciprofloxacin',
        'meropenem': 'Meropenem',
        'vancomycin': 'Vancomycin',
        'amikacin': 'Amikacin'
    }

    available_abx = {k: v for k, v in antibiotics.items() if k in df.columns}

    if len(available_abx) == 0:
        print("No antibiotic treatment data available")
        return None

    # Get top microorganisms
    microbe_counts = df['microorganisms_clean'].value_counts()
    top_microbes = microbe_counts.head(6).index.tolist()

    # Create treatment matrix (what antibiotics were GIVEN to patients with each pathogen)
    treatment_matrix = []
    for microbe in top_microbes:
        microbe_data = df[df['microorganisms_clean'] == microbe]
        row = {'Pathogen': microbe, 'N': len(microbe_data)}
        for abx_code, abx_name in available_abx.items():
            if abx_code in microbe_data.columns:
                treated = (microbe_data[abx_code] == 'yes').sum()
                total = len(microbe_data)
                treatment_pct = (treated / total * 100) if total > 0 else 0
                row[abx_name] = f"{treatment_pct:.1f}%"
        treatment_matrix.append(row)

    treatment_df = pd.DataFrame(treatment_matrix)
    print("\n💊 Antibiotic Treatment Patterns (WHAT WAS PRESCRIBED):")
    print("="*60)
    print("NOTE: This shows the percentage of patients who RECEIVED each antibiotic,")
    print("NOT laboratory susceptibility testing results.")
    print("="*60)
    print(treatment_df.to_string(index=False))

    # Identify most commonly prescribed antibiotics
    print("\n🔬 MOST COMMONLY PRESCRIBED ANTIBIOTICS:")
    for abx_name in list(available_abx.values())[:5]:
        if abx_name in treatment_df.columns:
            # Extract percentages and calculate average
            percentages = []
            for microbe in top_microbes:
                val = treatment_df.loc[treatment_df['Pathogen'] == microbe, abx_name].values
                if len(val) > 0 and val[0] != '0.0%':
                    pct = float(val[0].replace('%', ''))
                    percentages.append(pct)
            if percentages:
                avg_use = sum(percentages) / len(percentages)
                print(f"  • {abx_name}: Prescribed in {avg_use:.1f}% of cases on average")

    # Create heatmap for treatment patterns
    fig, ax = plt.subplots(figsize=(14, 8))
    heatmap_data = []
    row_labels = []
    col_labels = list(available_abx.values())

    for microbe in top_microbes:
        microbe_data = df[df['microorganisms_clean'] == microbe]
        row = []
        for abx_code, abx_name in available_abx.items():
            if abx_code in microbe_data.columns:
                treated = (microbe_data[abx_code] == 'yes').sum()
                total = len(microbe_data)
                treatment_pct = (treated / total * 100) if total > 0 else 0
                row.append(treatment_pct)
            else:
                row.append(0)
        heatmap_data.append(row)
        row_labels.append(microbe)

    im = ax.imshow(heatmap_data, cmap='Blues', aspect='auto', vmin=0, vmax=100)
    ax.set_xticks(range(len(col_labels)))
    ax.set_xticklabels(col_labels, rotation=45, ha='right', fontsize=9)
    ax.set_yticks(range(len(row_labels)))
    ax.set_yticklabels(row_labels, fontsize=10)
    plt.colorbar(im, ax=ax, label='Patients Who Received Antibiotic (%)')

    for i in range(len(heatmap_data)):
        for j in range(len(heatmap_data[0])):
            if heatmap_data[i][j] > 0:
                ax.text(j, i, f'{heatmap_data[i][j]:.0f}%',
                       ha="center", va="center",
                       color="white" if heatmap_data[i][j] > 50 else "black",
                       fontsize=8, fontweight='bold')

    ax.set_title('Antibiotic Treatment Patterns by Pathogen', fontsize=12, fontweight='bold')
    plt.tight_layout()
    plt.savefig('Figure_5_Treatment_Patterns.png', dpi=300, bbox_inches='tight')
    plt.show()

    print("\n✅ SPECIFIC OBJECTIVE 3 (Part 2) COMPLETED: Treatment patterns described")
    print("✅ RESEARCH QUESTION 4 ANSWERED: Antibiotic treatment patterns identified")

    return treatment_df

# Run the corrected analysis
treatment_results = antibiotic_treatment_analysis(df)
```

```python
# ============================================
# SECTION 10: TREATMENT TRENDS ANALYSIS
# ============================================

print("\n" + "="*80)
print("SECTION 10: TREATMENT TRENDS ANALYSIS")
print("="*80)
print("🎯 SPECIFIC OBJECTIVE 3: To describe treatment given")
print("📋 RESEARCH QUESTION 5: Are there changes in type of treatment given?")
print("="*80)

def treatment_trends_analysis(df):
    """Analyze treatment patterns over time"""

    treatment_vars = {}
    if 'iv antibiotics with operation' in df.columns:
        treatment_vars['IV Antibiotics with Operation'] = 'iv antibiotics with operation'
    if 'wound care and oral antibiotics' in df.columns:
        treatment_vars['Wound Care + Oral Antibiotics'] = 'wound care and oral antibiotics'
    if 'prophylaxis antibiotic' in df.columns:
        treatment_vars['Prophylaxis Antibiotics'] = 'prophylaxis antibiotic'

    if len(treatment_vars) == 0:
        print("No treatment data available")
        return None

    treatment_trends = []
    for year in sorted(df['year'].unique()):
        year_data = df[df['year'] == year]
        row = {'Year': year, 'N': len(year_data)}
        for treatment_name, treatment_col in treatment_vars.items():
            if treatment_col in df.columns:
                treatment_count = (year_data[treatment_col] == 'yes').sum()
                treatment_pct = (treatment_count / len(year_data) * 100) if len(year_data) > 0 else 0
                row[treatment_name] = treatment_pct
        treatment_trends.append(row)

    treatment_df = pd.DataFrame(treatment_trends)
    print("\n💊 Treatment Trends Over Time:")
    print(treatment_df.to_string(index=False))

    # Identify changes in treatment
    print("\n📈 TREATMENT PATTERN CHANGES:")
    for treatment_name in treatment_vars.keys():
        if treatment_name in treatment_df.columns:
            first_pct = treatment_df[treatment_name].iloc[0]
            last_pct = treatment_df[treatment_name].iloc[-1]
            change = last_pct - first_pct
            direction = "increased" if change > 0 else "decreased"
            print(f"  • {treatment_name}: {first_pct:.1f}% (2014) → {last_pct:.1f}% (2023) ({direction} by {abs(change):.1f}%)")

    # Create figure
    fig, ax = plt.subplots(figsize=(10, 6))
    for treatment_name in treatment_vars.keys():
        if treatment_name in treatment_df.columns:
            ax.plot(treatment_df['Year'], treatment_df[treatment_name],
                   'o-', linewidth=2, markersize=8, label=treatment_name)

    ax.set_xlabel('Year', fontsize=11)
    ax.set_ylabel('Percentage of Patients (%)', fontsize=11)
    ax.set_title('Treatment Trends in FRI Patients (2014-2023)', fontsize=12, fontweight='bold')
    ax.legend(loc='best')
    ax.grid(True, alpha=0.3)
    ax.set_xticks(range(2014, 2024))
    ax.set_ylim(0, 105)
    plt.tight_layout()
    plt.savefig('Figure_6_Treatment_Trends.png', dpi=300, bbox_inches='tight')
    plt.show()

    print("\n✅ SPECIFIC OBJECTIVE 3 (Part 3) COMPLETED: Treatment patterns described")
    print("✅ RESEARCH QUESTION 5 ANSWERED: Changes in treatment identified")

    return treatment_df

treatment_results = treatment_trends_analysis(df)

```
